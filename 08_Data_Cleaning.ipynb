<h1>Male Names</h1>

```python
import pandas as pd
```

```python
def male_names():
    
    data = pd.read_csv("Indian-Male-Names.csv")
    data['first_name'] = male_names['name'].str.split().str[0]
    newdata = male_names.drop(columns=["gender" , "name" , "race"])
    newdata.to_csv("male_first_names.csv" , index=False)
```

<h1>female names</h1>

```python
def female_names():
    
    data = pd.read_csv("Indian-Female-Names.csv")
    data["first_name"] = data["name"].str.split().str[0]
    new_data = data.drop(columns=["name" , "gender" , "race"])
    new_data.to_csv("Female_fist_names.csv" , index= False)
```
