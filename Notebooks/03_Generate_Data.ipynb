<h2>genrate name</h2>

```python
def generate_name():

    gender = np.random.choice(
        ["Male", "Female"],
        p=[0.6, 0.4]
    )

    if gender == "Male":
        first_name = random.choice(male_names["first_name"].tolist())
    else:
        first_name = random.choice(female_names["first_name"].tolist())

    last_name = random.choice(last_names["caste"].tolist())

    full_name = first_name + " " + last_name

    return str(gender), full_name
```

<h2>generate DOB</h2>

```python
import numpy as np
from datetime import date, timedelta

def generate_dob():

    today = date.today()

    # 18 to 80 years old
    start_date = date(today.year - 80, 1, 1)
    end_date = date(today.year - 18, 12, 31)

    # Total days in the range
    total_days = (end_date - start_date).days

    # Pick a random day
    random_days = np.random.randint(0, total_days + 1)

    dob = start_date + timedelta(days=random_days)
    return dob
```

<h2>generate age</h2>

```python
from datetime import date

def calculate_age(dob):

    today = date.today()

    age = today.year - dob.year

    if (today.month, today.day) < (dob.month, dob.day):
        age -= 1

    return age
```

<h1>generate mobile number</h1>

```python
import numpy as np

def generate_mobile_number():

    first_digit = np.random.choice([6, 7, 8, 9])

    remaining_digits = "".join(
        np.random.choice(list("0123456789"), size=9)
    )

    mobile = str(first_digit) + remaining_digits

    return mobile
```

<h1>generate email</h1>

```python
def generate_email(name):
    name = name.replace(" ","").lower()
    middle = "".join(np.random.choice(list("012345678"), size=2))

    return f"{name}{middle}@gmail.com"
```

<h1>generate_account_opening_date</h1>

```python
from datetime import date, timedelta
import numpy as np

def generate_account_opening_date(dob):

    today = date.today()

    try:
        eligible_date = date(
            dob.year + 18,
            dob.month,
            dob.day
        )
    except ValueError:
        eligible_date = date(
            dob.year + 18,
            dob.month,
            28
        )

    if eligible_date >= today:
        return today

    total_days = (today - eligible_date).days

    random_days = np.random.randint(0, total_days + 1)

    opening_date = eligible_date + timedelta(days=random_days)

    return opening_date
```

<h1>Now The final one Customer generation</h1>

```python
def generate_customer():

    gender , full_name = generate_name()

    dob = generate_dob()
    age = calculate_age(dob)

    city, state = generate_city_state()

    mobile = generate_mobile_number()
    email = generate_email(full_name)

    pan = generate_pan_number()
    aadhaar = generate_aadhaar_number()

    marital_status = generate_marital_status(age)
    occupation = generate_occupation()

    income = generate_income()
    cibil_score = generate_cibil_score()

    account_type = generate_account_type()
    account_number = generate_account_number()
    ifsc_code = generate_ifsc_code()
    branch_name = generate_branch_name()

    account_opening_date = generate_account_opening_date(dob)
    account_balance = generate_account_balance()

    loan_status = generate_loan_status()

    return {
        "full_name": full_name,
        "gender": gender,
        "dob": dob,
        "age": age,
        "city": city,
        "state": state,
        "mobile": mobile,
        "email": email,
        "pan_number": pan,
        "aadhaar_number": aadhaar,
        "marital_status": marital_status,
        "occupation": occupation,
        "annual_income": income,
        "cibil_score": cibil_score,
        "account_type": account_type,
        "account_number": account_number,
        "ifsc_code": ifsc_code,
        "branch_name": branch_name,
        "account_opening_date": account_opening_date,
        "account_balance": account_balance,
        "loan_status": loan_status
    }
```

<H1>Now we make new Table ( Account Table ) </H1>

```python
import pandas as pd
from sqlalchemy import create_engine , text
from sqlalchemy.engine import URL

from dotenv import load_dotenv
import os

load_dotenv()

def get_engine():
    url = URL.create(
        drivername="mysql+pymysql",
        username=os.getenv("DB_USER"),
        password=os.getenv("DB_PASSWORD"),
        host=os.getenv("DB_HOST"),
        port=3306,
        database="Banking_analytics"
    )
    engine = create_engine(url)

    return engine
```

```python
engine = get_engine()
query = """
        CREATE TABLE IF NOT EXISTS accounts (

        account_id INT AUTO_INCREMENT PRIMARY KEY,
    
        customer_id INT NOT NULL,
    
        account_number BIGINT UNIQUE NOT NULL,
    
        account_type VARCHAR(20) NOT NULL,
    
        ifsc_code VARCHAR(20) NOT NULL,
    
        branch_name VARCHAR(100) NOT NULL,
    
        account_opening_date DATE NOT NULL,
    
        account_balance DECIMAL(15,2) NOT NULL,
    
        FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id)
    
    );
    """
with engine.begin()as conn:
    conn.execute(text(query))

print("table created succesfully")
```
