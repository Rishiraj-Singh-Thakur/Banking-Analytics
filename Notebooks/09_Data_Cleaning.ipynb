<h1>data correction for ML Algorithms</h1>

```python
from sqlalchemy import create_engine,text
from sqlalchemy.engine import URL

from dotenv import load_dotenv
import os

load_dotenv()


url = URL.create(
        drivername="mysql+pymysql",
        username=os.getenv("DB_USER"),
        password=os.getenv("DB_PASSWORD"),
        host=os.getenv("DB_HOST"),
        port=3306,
        database="Banking_analytics"
    )
engine = create_engine(url)
```

```python
import pandas as pd
customers = pd.read_sql("customers", engine)
loans = pd.read_sql("loans" , engine)
accounts = pd.read_sql("accounts",engine)
transactions = pd.read_sql("transactions",engine)

print("CUSTOMERS")
print(customers.columns.tolist())

print("\nACCOUNTS")
print(accounts.columns.tolist())

print("\nLOANS")
print(loans.columns.tolist())

print("\nTRANSACTIONS")
print(transactions.columns.tolist())
```

```python
import numpy as np

r = loans["interest_rate"] / 100 / 12
n = loans["tenure_months"]
P = loans["loan_amount"]

calculated_emi = (
    P * r * (1 + r) ** n
) / (
    (1 + r) ** n - 1
)

emi_difference = abs(loans["emi"] - calculated_emi)

emi_difference.describe()
```
