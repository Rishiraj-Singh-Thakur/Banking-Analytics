<h2>Create customer table</h2>

```python
from sqlalchemy import text ,create_engine
from sqlalchemy.engine import URL
```

```python
def get_engine():
    url = URL.create(
        drivername = "mysql+pymysql",
        username = "root",
        password = "Mysql@123",
        host = "localhost",
        port = 3306,
        database="Banking_Analytics"
    )
    return create_engine(url)
```

```python
def create_customers_table(engine):
    query = """
    CREATE TABLE IF NOT EXISTS customers (

        customer_id INT AUTO_INCREMENT PRIMARY KEY,

        first_name VARCHAR(50) NOT NULL,
        last_name VARCHAR(50) NOT NULL,

        gender ENUM('Male','Female','Other') NOT NULL,

        date_of_birth DATE NOT NULL,
        age INT NOT NULL,

        mobile_number VARCHAR(15) UNIQUE,
        email VARCHAR(100) UNIQUE,

        city VARCHAR(50) NOT NULL,
        state VARCHAR(50) NOT NULL,
        pincode VARCHAR(6) NOT NULL,

        occupation VARCHAR(50),

        employment_type ENUM(
            'Salaried',
            'Self-Employed',
            'Business',
            'Student',
            'Retired'
        ),

        annual_income DECIMAL(12,2),
        monthly_income DECIMAL(12,2),

        account_balance DECIMAL(15,2) DEFAULT 0,

        cibil_score INT,

        account_type ENUM(
            'Savings',
            'Current',
            'Salary'
        ),

        account_status ENUM(
            'Active',
            'Inactive',
            'Closed'
        ) DEFAULT 'Active',

        debit_card BOOLEAN DEFAULT FALSE,
        credit_card BOOLEAN DEFAULT FALSE,
        internet_banking BOOLEAN DEFAULT FALSE,
        mobile_banking BOOLEAN DEFAULT FALSE,

        transaction_count INT DEFAULT 0,
        avg_monthly_transaction DECIMAL(15,2),

        customer_since DATE,

        last_transaction_date DATE,

        risk_category ENUM(
            'Low',
            'Medium',
            'High'
        )

    );
    """

    with engine.connect() as conn:
        conn.execute(text(query))
        conn.commit()

    print("✅ Customers Table Created Successfully")
```

```python
# call funtion
engine = get_engine()
create_customers_table(engine)
```
